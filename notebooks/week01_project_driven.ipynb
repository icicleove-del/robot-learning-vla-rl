# Day 2 · 追踪 SmolVLA 推理与动作块

**核心问题：** 为什么策略可能预测一段动作，而环境通常按时间步接受动作？

**先看真实项目：**
- [`modeling_smolvla.py`](https://github.com/huggingface/lerobot/blob/v0.6.1/src/lerobot/policies/smolvla/modeling_smolvla.py)：搜索 `select_action`、`predict_action_chunk`，分别记录输入、输出、队列/缓存相关变量。
- [`configuration_smolvla.py`](https://github.com/huggingface/lerobot/blob/v0.6.1/src/lerobot/policies/smolvla/configuration_smolvla.py)：查询 action chunk 与模型配置项，记录真实版本中的变量名与默认值；**不要凭教程臆造参数**。

**按需补理论：** Python 类、对象状态、tensor shape、时间维 `[T, action_dim]`、Action Chunking 与闭环重新规划。

**2.5 小时安排：** 20 分钟看文件结构 → 70 分钟追 `select_action` → 40 分钟写动作块的最小示例 → 20 分钟归纳。

**交付：** `select_action → predict_action_chunk → action queue → environment step` 的函数调用图，标出**真实函数与玩具模型之间的区别**。

**停止条件：** 能解释「一次预测多少步 / 每次执行多少步」是两个不同配置问题；若某环节难以确认，标记未核实而不是默认它必然使用某种执行策略。

````python
# Day 2 · 纯 Python 解释动作块消费过程（教学示例，不是 SmolVLA 内部实现）
from collections import deque

class ToyChunkExecutor:
    def __init__(self, chunk_size=3):
        if chunk_size <= 0:
            raise ValueError("chunk_size 必须为正整数")
        self.chunk_size = chunk_size
        self.pending = deque()
        self.predict_calls = 0

    def predict_chunk(self, observation):
        self.predict_calls += 1
        # 每次给出相同维度、不同时间步的三个示例动作
        return [[observation["x"] + 0.1 * i] for i in range(self.chunk_size)]

    def select_action(self, observation):
        if not self.pending:
            self.pending.extend(self.predict_chunk(observation))
        return self.pending.popleft()

policy = ToyChunkExecutor(chunk_size=3)
for step in range(5):
    action = policy.select_action({"x": float(step)})
    print(f"environment_step={step}, action={action}, model_calls={policy.predict_calls}")
assert policy.predict_calls == 2  # 5 个环境步、每块 3 步 => 共预测 2 次
````

# Day 4 · 看懂 SmolVLA 的 forward 与 Flow Matching 损失

**核心问题：** 为什么从带噪动作出发，就能构造训练目标？

**先看真实项目：**
- [`modeling_smolvla.py`](https://github.com/huggingface/lerobot/blob/v0.6.1/src/lerobot/policies/smolvla/modeling_smolvla.py)：定位 `forward`，追踪训练动作、噪声、插值时间和训练 loss；记录源码中的 shape、mask 和 reduction。
- 暂时只掌握与动作预测相关的部分，Transformer 细节只补到理解具体函数输入和输出为止。

**最小核心概念（需回到源码核对具体符号约定）：**

$$x_t = (1-t)a+t\epsilon, \qquad v_{target}=\epsilon-a$$

- $a$：示范动作；$\epsilon$：采样噪声；$t$：插值时间；$x_t$：用于训练的中间动作。
- 模型根据输入条件与 $t$ 预测速度场，并与目标速度比较形成损失。
- **注意：** 这是理解 Flow Matching 的简化写法；真实源码还可能包含 mask、时间采样分布、动作 padding 等逻辑。

**2.5 小时安排：** 60 分钟读 `forward` → 40 分钟自己解释/计算简化表达式 → 30 分钟按需补 PyTorch/Flow Matching → 20 分钟给出从源码到公式的逐项对应。

**交付：** 一页张量流向图 + 每个变量的语义和形状。

**停止条件：** 能解释「预测的到底是什么」「为什么损失能提供监督」，而不是只复述某个公式。

````python
# Day 4 · 标量版 Flow Matching 自检（Python 标准库，非训练）
# 只用于理解插值和监督目标。真实模型预测的是高维动作速度场。
a, epsilon, t = 0.2, -0.8, 0.4
x_t = (1 - t) * a + t * epsilon
v_target = epsilon - a
# 假设一个教学用固定预测值，观察平方误差；未做模型训练。
v_prediction = -0.7
loss = (v_prediction - v_target) ** 2
print({"clean_action": a, "noise": epsilon, "time": t,
       "interpolated_action": x_t, "target_velocity": v_target, "toy_mse": loss})
assert abs(x_t - (-0.2)) < 1e-9
assert abs(v_target - (-1.0)) < 1e-9
````

- 与 Day 4 的 `forward` 对照，分清“模型负责计算 loss”和“训练脚本负责参数更新”。

**按需补理论：** 计算图、梯度、自动求导、learning rate、zero_grad、梯度累积；不要求先修完整优化理论。

**2.5 小时安排：** 75 分钟追训练入口 → 35 分钟写简化参数更新解释 → 20 分钟按需补 autograd → 20 分钟复盘调用链。

**交付：** `batch → policy.forward → loss → backward → optimizer.step → checkpoint/log` 的真实源码定位表。

**停止条件：** 说明一次参数更新包含哪些条件；不得把一次 loss 下降理解成 LIBERO 成功率提高。

````python
# Day 5 · 选做的 CPU 最小梯度观察（需本地已安装 PyTorch，手动执行；无下载与训练大模型）
# 未安装 PyTorch 时只输出说明，不触发 pip install。
from importlib.util import find_spec
if find_spec("torch") is None:
    print("本环境没有 PyTorch。本周可以先完成源码阅读；不要为了运行此示例擅自安装依赖。")
else:
    import torch
    w = torch.tensor(0.5, requires_grad=True)
    x = torch.tensor(2.0)
    y_true = torch.tensor(4.0)
    prediction = w * x
    loss = (prediction - y_true) ** 2
    loss.backward()
    print({"prediction": prediction.item(), "loss": loss.item(), "dLoss/dw": w.grad.item()})
    # 这是手算的单次教学更新演示，不涉及预训练模型、任务或数据集。
    with torch.no_grad():
        w_next = w - 0.1 * w.grad
    print("如果沿负梯度做一次小步更新，w 将从", w.item(), "变为", w_next.item())
````

**按需补理论：** `deque`、面向对象状态、边界条件测试、时间步同步；不会什么补什么。

````python
# Day 6 · 先自己写，再参考下面的检查题（这里只保留可运行的最小脚手架）
from collections import deque

class MinimalActionQueue:
    """教学用动作队列。不是官方 SmolVLA 实现。"""
    def __init__(self):
        self.queue = deque()

    def push_chunk(self, actions):
        if self.queue:
            raise RuntimeError("尚有未消费动作，先明确该如何处理")
        if not actions:
            raise ValueError("动作块不能为空")
        self.queue.extend(actions)

    def next_action(self):
        if not self.queue:
            raise IndexError("没有可执行的动作")
        return self.queue.popleft()

# ① 正常使用 ② 空动作块 ③ 耗尽后再次消费
q = MinimalActionQueue()
q.push_chunk([[0.1, 0.2], [0.3, 0.4]])
assert q.next_action() == [0.1, 0.2]
assert q.next_action() == [0.3, 0.4]
try:
    q.next_action()
except IndexError:
    print("测试通过：队列耗尽会抛出 IndexError")
else:
    raise AssertionError("队列耗尽没有正确报错")

try:
    MinimalActionQueue().push_chunk([])
except ValueError:
    print("测试通过：禁止空动作块")
else:
    raise AssertionError("空动作块没有正确报错")
````
